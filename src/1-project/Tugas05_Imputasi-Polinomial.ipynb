### 2. Penanganan Missing Values: Polinomial Orde 2 dan Fallback

Untuk mengisi nilai yang hilang tanpa membuang tanggal observasi, interpolasi polinomial orde 2 (`interpolate(method="polynomial", order=2)`) diterapkan secara terpisah pada CO, NO2, SO2, dan CH4. Metode ini memperkirakan nilai berdasarkan pola kurva dari observasi valid di sekitarnya.

#### Formulasi Umum Interpolasi Polinomial:
$$
\hat{x}(t) = \sum_{i=0}^{n} x_i \prod_{\substack{j=0 \\ j \ne i}}^{n} \frac{t-t_j}{t_i-t_j}
$$

Interpolasi polinomial orde 2 hanya dapat diterapkan jika tersedia sedikitnya tiga observasi valid. Untuk nilai hilang di antara observasi valid, estimasi dihitung dari interpolasi kuadratik. Nilai hilang di awal atau akhir deret diisi dengan observasi valid terdekat (`bfill`/`ffill`). Jika jumlah observasi valid kurang dari tiga, interpolasi polinomial tidak dapat dihitung dan semua nilai hilang diisi dengan observasi valid terdekat. Jika seluruh seri tidak memiliki nilai valid, imputasi tidak dapat dilakukan.

**Catatan dataset:** CH4 memiliki 1 nilai valid dan 364 nilai hilang dari 365 baris. Karena itu, nilai CH4 yang hilang menggunakan fallback observasi valid terdekat; hasil CH4 menjadi nilai konstan, bukan estimasi polinomial. Ringkasan kode di bawah mencatat metode yang berlaku untuk setiap polutan.

- $\hat{x}(t)$: nilai konsentrasi polutan hasil estimasi pada waktu $t$.
- $x_i$: nilai observasi valid pada waktu $t_i$.
- $n$: orde polinomial interpolasi.

```python
# Gunakan interpolasi polinomial orde 2; fallback ke observasi valid terdekat bila perlu.
def polynomial_impute(series, order=2):
    values = pd.to_numeric(series, errors="coerce").replace(
        [np.inf, -np.inf], np.nan
    )
    valid_count = int(values.notna().sum())
    if valid_count == 0:
        return values
    if valid_count < order + 1:
        return values.bfill().ffill()
    return values.interpolate(method="polynomial", order=order).bfill().ffill()


pollutants = ["CO", "NO2", "SO2", "CH4"]
df_imputed = df_raw[["DATE_TIME"]].copy()
imputation_summary = []

for pollutant in pollutants:
    original = pd.to_numeric(df_raw[pollutant], errors="coerce")
    imputed = polynomial_impute(original, order=2)
    df_imputed[f"{pollutant}_imputed"] = imputed
    valid_count = int(original.notna().sum())
    if valid_count < 3:
        method = "Fallback: observasi valid terdekat (kurang dari 3 nilai valid)"
    elif original.isna().any():
        method = "Polinomial orde 2 + fallback nilai terdekat di ujung deret"
    else:
        method = "Tidak perlu imputasi (tidak ada missing value)"
    imputation_summary.append(
        {
            "Polutan": pollutant,
            "Nilai valid": valid_count,
            "Missing sebelum": int(original.isna().sum()),
            "Missing sesudah": int(imputed.isna().sum()),
            "Metode": method,
        }
    )

imputation_summary = pd.DataFrame(imputation_summary).set_index("Polutan")
if imputation_summary["Missing sesudah"].any():
    raise ValueError(
        "Imputasi menyisakan missing values: "
        f"{imputation_summary['Missing sesudah'].to_dict()}"
    )

# Pertahankan nama ini untuk langkah-langkah CO yang menyusul.
co_imputed = co_raw.copy()
co_imputed["CO_imputed"] = df_imputed["CO_imputed"]

print("Hasil Imputasi per Polutan (Polinomial Orde 2 dengan Fallback):")
print(imputation_summary.to_string())
```

### 3. Identifikasi dan Penanganan Outlier (Interquartile Range Capping / Winsorization)

Untuk mendeteksi dan membatasi pencilan ekstrem tanpa membuang tanggal observasi, digunakan metode **Interquartile Range (IQR)** yang dilanjutkan dengan **Winsorization Capping**.

#### Formulasi Matematika Metode IQR:
$$\text{IQR} = Q_3 - Q_1$$
$$\text{Batas Bawah (Lower Bound)} = Q_1 - 1{,}5 \times \text{IQR}$$
$$\text{Batas Atas (Upper Bound)} = Q_3 + 1{,}5 \times \text{IQR}$$

#### Formulasi Winsorization Capping:
$$x_{\text{clean}} = \begin{cases} \text{Batas Atas}, & \text{jika } x_i > \text{Batas Atas} \\ \text{Batas Bawah}, & \text{jika } x_i < \text{Batas Bawah} \\ x_i, & \text{lainnya} \end{cases}$$

#### Keterangan Simbol Rumus:
- $Q_1$ : Kuartil Pertama (Persentil ke-25 data polutan ter-imputasi).
- $Q_3$ : Kuartil Ketiga (Persentil ke-75 data polutan ter-imputasi).
- $\text{IQR}$ : Jangkauan Interkuartil ($Q_3 - Q_1$).
- $1{,}5$ : Pengali standar baku Tukey untuk mendeteksi pencilan moderat.
- $x_{\text{clean}}$ : Nilai konsentrasi polutan bersih setelah dibatasi (*capped*).

#### Cara Membaca Rumus:
> *"Batas toleransi outlier dihitung secara terpisah untuk setiap polutan menggunakan IQR. Nilai yang melampaui batas atas atau bawah dibatasi agar tidak merusak distribusi data."*


```python
import numpy as np
import pandas as pd

# Terapkan IQR capping secara terpisah pada keempat polutan yang sudah diimputasi.
df_clean_all = df_imputed[["DATE_TIME"]].copy()
outlier_summary = []

for pollutant in pollutants:
    imputed_column = f"{pollutant}_imputed"
    values = df_imputed[imputed_column]
    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers_detected = (values < lower_bound) | (values > upper_bound)
    df_clean_all[f"{pollutant}_clean"] = np.clip(
        values, lower_bound, upper_bound
    )
    outlier_summary.append(
        {
            "Polutan": pollutant,
            "Q1": q1,
            "Q3": q3,
            "IQR": iqr,
            "Batas bawah": lower_bound,
            "Batas atas": upper_bound,
            "Jumlah outlier": int(outliers_detected.sum()),
        }
    )

# Pertahankan nama variabel CO yang digunakan pada langkah ekstraksi fitur.
co_clean = co_imputed.copy()
co_clean["pollutant_clean"] = df_clean_all["CO_clean"]

outlier_summary = pd.DataFrame(outlier_summary).set_index("Polutan")
print("Hasil Deteksi dan Perbaikan Outlier (IQR Capping) per Polutan:")
print(outlier_summary.to_string(float_format=lambda value: f"{value:.6f}"))
```

### 5. Analisis Kemiripan Sinyal Polutan (Similarity Analysis)

Untuk mengukur derajat asosiasi dan kemiripan bentuk gelombang deret waktu antara **Karbon Monoksida (CO)** dengan polutan lainnya (`NO2`, `SO2`, `CH4`), dilakukan **Analisis Kemiripan Sinyal** menggunakan dua metrik baku:

#### 1. Koefisien Korelasi Pearson ($r$):
$$r_{XY} = \frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum_{i=1}^{n} (x_i - \bar{x})^2} \cdot \sqrt{\sum_{i=1}^{n} (y_i - \bar{y})^2}}$$

#### 2. Cosine Similarity ($S_C$):
$$S_C(\mathbf{A}, \mathbf{B}) = \frac{\mathbf{A} \cdot \mathbf{B}}{\|\mathbf{A}\| \|\mathbf{B}\|} = \frac{\sum_{i=1}^{n} A_i B_i}{\sqrt{\sum_{i=1}^{n} A_i^2} \cdot \sqrt{\sum_{i=1}^{n} B_i^2}}$$

#### Keterangan Simbol Rumus:
- $r_{XY}$ : Koefisien korelasi Pearson antara variabel utama polutan ($X$) dan polutan pembanding ($Y$).
- $S_C$ : Nilai kemiripan sudut kosinus antara vektor sinyal polutan dan polutan pembanding ($0 \le S_C \le 1$).
- $\mathbf{A}, \mathbf{B}$ : Vektor deret waktu konsentrasi polutan harian.

#### Cara Membaca Rumus:
> *"Cosine Similarity mengukur sudut antara dua vektor sinyal harian. Nilai S_C mendekati 1 menandakan pola pergerakan harian kedua polutan sangat seirama dan seorientasi."*


```python
import numpy as np
from numpy.linalg import norm
import pandas as pd

# Gunakan data_bangkalan.xlsx yang sudah dimuat pada df_raw.
# Semua polutan disejajarkan berdasarkan tanggal yang sama.
comparison_columns = ["CO", "NO2", "SO2", "CH4"]
df_all = df_raw[["DATE_TIME"] + comparison_columns].copy()
for column in comparison_columns:
    df_all[column] = polynomial_impute(df_all[column], order=2)

missing_by_pollutant = df_all[comparison_columns].isna().sum()
if missing_by_pollutant.any():
    raise ValueError(f"Masih ada missing value: {missing_by_pollutant.to_dict()}")

pearson_corr = df_all[comparison_columns].corr()["CO"]


def cosine_sim(a, b):
    denominator = norm(a) * norm(b)
    return float(np.dot(a, b) / denominator) if denominator else np.nan


cosine_sims = {}
co_vec = df_all["CO"].to_numpy(dtype=float)
for column in comparison_columns:
    cosine_sims[column] = cosine_sim(co_vec, df_all[column].to_numpy(dtype=float))

similarity_df = pd.DataFrame(
    {
        "Parameter Polutan": comparison_columns,
        "Pearson Correlation (r)": [pearson_corr[column] for column in comparison_columns],
        "Cosine Similarity (Sc)": [cosine_sims[column] for column in comparison_columns],
    }
)

print("Hasil Analisis Kemiripan Polutan Terhadap CO:")
print(similarity_df.to_string(index=False))
print("Missing values setelah imputasi:")
print(missing_by_pollutant.to_string())
```

```python
import os
import numpy as np
import pandas as pd
import tsfel

# Gunakan workbook sumber yang sama dengan sel sebelumnya.
PROJECT_ROOT = get_project_root()
DATA_DIR = os.path.join(PROJECT_ROOT, "data")
file_path = os.path.join(DATA_DIR, "data_polutan_bangkalan.xlsx")

source = pd.read_excel(file_path, sheet_name="Data Harian")
column_map = {
    "Tanggal": "DATE_TIME",
    "NO2 (Harian)": "NO2",
    "CO (Harian)": "CO",
    "SO2 (Harian)": "SO2",
    "CH4 (Harian)": "CH4",
}
missing_columns = [column for column in column_map if column not in source.columns]
if missing_columns:
    raise ValueError(f"Kolom wajib tidak ditemukan: {missing_columns}")

df_raw = source[list(column_map)].rename(columns=column_map)
df_raw["DATE_TIME"] = pd.to_datetime(df_raw["DATE_TIME"], errors="coerce")
if df_raw["DATE_TIME"].isna().any():
    raise ValueError("Kolom Tanggal mengandung nilai yang tidak valid")

pollutants = ["CO", "SO2", "CH4", "NO2"]
for pol in pollutants:
    df_raw[pol] = pd.to_numeric(df_raw[pol], errors="coerce")
df_raw = (
    df_raw.sort_values("DATE_TIME")
    .drop_duplicates("DATE_TIME")
    .reset_index(drop=True)
)

# Ekstraksi per domain mencegah konflik nama fitur pada TSFEL versi terbaru.
cfg_stat = tsfel.get_features_by_domain("statistical")
cfg_temp = tsfel.get_features_by_domain("temporal")
cfg_spec = tsfel.get_features_by_domain("spectral")


def extract_tsfel_features(signal):
    results = [
        tsfel.time_series_features_extractor(cfg_stat, signal, fs=1, verbose=0),
        tsfel.time_series_features_extractor(cfg_temp, signal, fs=1, verbose=0),
    ]
    for feature_name, feature_params in cfg_spec["spectral"].items():
        if feature_name in {
            "Wavelet absolute mean",
            "Wavelet energy",
            "Wavelet standard deviation",
            "Wavelet variance",
        }:
            continue
        try:
            result = tsfel.time_series_features_extractor(
                {"spectral": {feature_name: feature_params}},
                signal,
                fs=1,
                verbose=0,
            )
            results.append(result)
        except Exception:
            continue

    features = pd.concat(results, axis=1)
    unique_columns = []
    seen = {}
    for column in features.columns:
        base_name = column.replace("0_", "")
        count = seen.get(base_name, 0)
        unique_columns.append(base_name if count == 0 else f"{base_name}_{count}")
        seen[base_name] = count + 1
    features.columns = unique_columns

    # Kurtosis dan skewness tidak terdefinisi untuk sinyal konstan seperti CH4.
    # Ubah NaN/inf pada fitur turunan menjadi angka netral agar CSV siap dipakai.
    features = features.apply(pd.to_numeric, errors="coerce")
    features = features.replace([np.inf, -np.inf], np.nan).fillna(0.0)
    return features


for pol in pollutants:
    print(f"\n================ MENGOLAH POLUTAN: {pol} ================")
    values = polynomial_impute(df_raw[pol], order=2)
    if values.isna().all():
        print(f"[WARNING] {pol} tidak memiliki nilai valid. Dilewati...")
        continue

    df_pol = pd.DataFrame(
        {"DATE_TIME": df_raw["DATE_TIME"], "pollutant_clean": values}
    )
    p1 = os.path.join(DATA_DIR, f"data_prepared_{pol.lower()}.csv")
    p2 = os.path.join(DATA_DIR, f"features_tsfel_{pol.lower()}.csv")
    df_pol.to_csv(p1, index=False)

    print(f"Sedang mengekstrak fitur TSFEL untuk {pol}...")
    features_pol = extract_tsfel_features(values.to_numpy(dtype=float))
    features_pol.to_csv(p2, index=False)
    non_finite_count = int(~np.isfinite(features_pol.to_numpy()).all())

    print(f"Berhasil menyimpan file untuk {pol}:")
    print(f"  - Prepared Data : {os.path.basename(p1)} ({len(df_pol)} baris)")
    print(f"  - TSFEL Features: {os.path.basename(p2)} ({features_pol.shape[1]} fitur)")
    print(f"  - Nilai NaN/inf pada fitur: {non_finite_count}")

print("\nStatus Data Preparation semua polutan: SELESAI")
```

```python
# Eksekusi imputasi polinomial orde 2 dengan fallback nilai terdekat.

import os
import numpy as np
import pandas as pd

polynomial_order = 2
pollutants = ["CO", "NO2", "SO2", "CH4"]

# Urutkan tanggal dan pastikan satu baris per tanggal.
df_polynomial = (
    df_raw.sort_values("DATE_TIME")
    .drop_duplicates("DATE_TIME")
    .reset_index(drop=True)
    .copy()
)


def polynomial_impute_with_dates(series, dates, order=2):
    values = pd.to_numeric(series, errors="coerce").replace(
        [np.inf, -np.inf], np.nan
    )
    valid_count = int(values.notna().sum())
    if valid_count == 0:
        return values
    if valid_count < order + 1:
        return values.bfill().ffill()

    # Gunakan jarak hari sebagai sumbu interpolasi.
    day_index = pd.DatetimeIndex(dates).map(pd.Timestamp.toordinal).to_numpy()
    dated_values = pd.Series(values.to_numpy(), index=day_index)
    imputed = dated_values.interpolate(
        method="polynomial",
        order=order,
        limit_area="inside",
    )

    # Polynomial tidak mengisi bagian awal/akhir deret; gunakan nilai terdekat
    # untuk missing value pada bagian tersebut.
    imputed = imputed.bfill().ffill()
    return pd.Series(imputed.to_numpy(), index=series.index)


for column in pollutants:
    original = pd.to_numeric(df_polynomial[column], errors="coerce").replace(
        [np.inf, -np.inf], np.nan
    )
    valid_count = int(original.notna().sum())
    missing_before = int(original.isna().sum())
    if valid_count == 0:
        raise ValueError(f"{column} tidak memiliki observasi valid untuk imputasi")
    df_polynomial[column] = polynomial_impute_with_dates(
        original,
        df_polynomial["DATE_TIME"],
        order=polynomial_order,
    )
    missing_after = int(df_polynomial[column].isna().sum())
    if valid_count < polynomial_order + 1:
        method = "fallback observasi valid terdekat (kurang dari 3 nilai valid)"
    elif missing_before:
        method = "polinomial orde 2 + fallback nilai terdekat di ujung deret"
    else:
        method = "tidak perlu imputasi"

    prepared_pol = df_polynomial[["DATE_TIME", column]].rename(
        columns={column: "pollutant_clean"}
    )
    output_path = os.path.join(
        DATA_DIR, f"data_prepared_imputed_{column.lower()}.csv"
    )
    prepared_pol.to_csv(output_path, index=False)

    print(
        f"{column}: nilai valid={valid_count}, metode={method}, "
        f"missing sebelum={missing_before}, sesudah={missing_after}; "
        f"tersimpan di {output_path}"
    )
```
