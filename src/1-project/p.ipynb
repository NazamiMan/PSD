# Digitasi label tutupan lahan

Setiap Feature harus berupa poligon satu kelas dengan properti `class_name`. Gunakan label `build_up`, `farm_land`, `water_body`, `forest`, atau `mangrove`. Tambahkan banyak poligon terpisah yang benar-benar mewakili tiap kelas; contoh di bawah hanya dua poligon dan belum cukup untuk melatih/evaluasi model lima kelas. Pastikan koordinat berada di wilayah AOI yang akan diproses.

```python
training_geojson = {
  "type": "FeatureCollection",
  "features": [
    {
      "type": "Feature",
      "geometry": {
        "type": "Polygon",
        "coordinates": [
          [[112.70, -7.25], [112.71, -7.25], [112.71, -7.26], [112.70, -7.26], [112.70, -7.25]]
        ]
      },
      "properties": {
        "id": 1,
        "class_name": "build_up",
        "description": "Area Permukiman"
      }
    },
    {
      "type": "Feature",
      "geometry": {
        "type": "Polygon",
        "coordinates": [
          [[112.20, -7.50], [112.21, -7.50], [112.21, -7.51], [112.20, -7.51], [112.20, -7.50]]
        ]
      },
      "properties": {
        "id": 2,
        "class_name": "farm_land",
        "description": "Lahan Sawah"
      }
    }
  ]
}
```

```python
from collections import Counter
import json
from pathlib import Path

labels = Counter(
    feature.get("properties", {}).get("class_name")
    for feature in training_geojson.get("features", [])
)
print("Jumlah poligon per kelas:", dict(labels))
missing = {"build_up", "farm_land", "water_body", "forest", "mangrove"} - set(labels)
if missing:
    print("Tambahkan kelas yang belum ada:", sorted(missing))
few = {label: count for label, count in labels.items() if count < 10}
if few:
    print("Disarankan menambah sampel (target awal ≥10 poligon/kelas):", few)

project_dir = Path.cwd()
if not (project_dir / "land_cover.py").exists() and (project_dir / "src" / "1-project" / "land_cover.py").exists():
    project_dir = project_dir / "src" / "1-project"
output_path = project_dir / "training-labels.geojson"
output_path.write_text(json.dumps(training_geojson, ensure_ascii=False, indent=2), encoding="utf-8")
print("GeoJSON tersimpan:", output_path.resolve())
```
