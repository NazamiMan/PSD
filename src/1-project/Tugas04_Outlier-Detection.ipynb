Penjelasan Alur Kerja (Workflow) Clustering pada KNIME

Dalam penelitian ini, proses pengelompokan (clustering) kualitas daerah dari 16 mahasiswa dilakukan menggunakan perangkat lunak KNIME. Dataset yang digunakan berisi hasil ekstraksi fitur Time Series Feature Extraction Library (TSFEL). Alur kerja dirancang untuk membandingkan hasil pengelompokan menggunakan dua pendekatan: pengelompokan langsung pada data mentah dan pengelompokan setelah dilakukan reduksi dimensi.

Berikut adalah rincian fungsi dari masing-masing node yang digunakan dalam alur kerja tersebut:

    1. Excel Reader Node ini berfungsi sebagai gerbang masukan (input) data. Tahapan ini bertugas membaca dan mengimpor dataset awal dari format Excel ke dalam lingkungan KNIME. Dataset yang dimuat terdiri dari 16 baris data (mewakili 16 daerah dari masing-masing mahasiswa) beserta seluruh variabel fitur kualitas daerah yang sebelumnya telah diekstraksi menggunakan TSFEL.

    2. Table View (Pengecekan Data Awal) Node ini terhubung langsung dengan Excel Reader dan berfungsi untuk menampilkan dataset mentah dalam bentuk tabel interaktif. Langkah ini merupakan bagian dari tahap pra-pemrosesan (preprocessing) untuk melakukan inspeksi visual, memastikan kelengkapan data, dan memvalidasi bahwa seluruh fitur TSFEL telah berhasil dimuat dengan benar sebelum dianalisis lebih lanjut.

    3. k-Means (Skenario 1: Tanpa Reduksi Dimensi) Node k-Means bagian atas menerima aliran data langsung dari Excel Reader. Node ini mengeksekusi algoritma clustering K-Means pada dataset asli secara utuh. Proses ini mengelompokkan 16 daerah tersebut berdasarkan seluruh dimensi fitur TSFEL yang ada tanpa melalui proses penyaringan atau reduksi fitur sebelumnya.

    4. PCA (Principal Component Analysis) Mengingat proses ekstraksi TSFEL sering kali menghasilkan dimensi fitur yang sangat besar (ratusan fitur), node PCA digunakan untuk melakukan reduksi dimensi. Algoritma ini mentransformasikan fitur-fitur TSFEL ke dalam sekumpulan variabel baru (principal components) yang lebih sedikit, tanpa menghilangkan informasi variansi data yang signifikan. Langkah ini krusial untuk mengatasi masalah curse of dimensionality, mengurangi noise, dan meningkatkan efisiensi komputasi pada algoritma clustering.

    5. Table View (Pengecekan Data Hasil PCA) Node ini terhubung pada output PCA dan berfungsi untuk menampilkan matriks data baru yang dimensinya telah berhasil direduksi. Tabel ini digunakan oleh peneliti untuk meninjau struktur nilai komponen utama sebelum data tersebut diteruskan ke tahap pemodelan.

    6. k-Means (Skenario 2: Dengan Reduksi Dimensi) Node k-Means pada bagian kanan bawah menerima data yang telah diproses oleh node PCA. Node ini menjalankan algoritma K-Means pada dataset yang lebih ringkas dan optimal.

Kesimpulan Alur Kerja Secara keseluruhan, arsitektur workflow bercabang ini sengaja dibangun untuk memfasilitasi evaluasi komparatif. Dengan membagi alur menjadi dua cabang, peneliti dapat membandingkan efektivitas dan kualitas hasil clustering daerah antara model yang menggunakan fitur TSFEL mentah secara keseluruhan dengan model yang menggunakan fitur hasil reduksi dimensi PCA.
